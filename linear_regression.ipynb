```python

# Step 1: Upload CSV file
from google.colab import files
uploaded = files.upload()

# Step 2: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Step 3: Read CSV file
df = pd.read_csv('regression.csv')

print(df)

# Step 4: Select X and Y
X = df[['Hours_Studied']]
y = df['Marks']

# Step 5: Create Linear Regression model
model = LinearRegression()

# Step 6: Train the model
model.fit(X, y)

# Step 7: Print model equation
b0 = model.intercept_
b1 = model.coef_[0]

print("\nLinear Regression Equation")
print(f"Marks = {b0:.4f} + {b1:.4f}*x")

# Step 8: Predict values
y_pred = model.predict(X)

print("\nPredicted Values:")
print(y_pred)

# Step 9: Calculate MSE
mse = mean_squared_error(y, y_pred)
print("\nTraining MSE:", round(mse, 4))

# Step 10: Calculate R2 Score
r2 = r2_score(y, y_pred)
print("R2 Score:", round(r2, 4))

# Step 11: Visualize the graph
plt.scatter(X, y, color='blue', label='Actual Data')

plt.plot(X, y_pred, color='red', label='Regression Line')

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Linear Regression")

plt.legend()
plt.show()
```
