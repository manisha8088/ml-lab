```python

# Step 1: Upload CSV file
from google.colab import files
uploaded = files.upload()

# Step 2: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Step 3: Read CSV file
df = pd.read_csv('regression.csv')
print(df)

# Step 4: Select X and Y
X = df[['Hours_Studied']]
y = df['Marks']

# Step 5: Create degree 2 polynomial features
poly2 = PolynomialFeatures(degree=2, include_bias=False)
X_poly2 = poly2.fit_transform(X)

# Step 6: Train model
model2 = LinearRegression()
model2.fit(X_poly2, y)

# Step 7: Print model equation
b0 = model2.intercept_
b1, b2 = model2.coef_

print("\nDegree 2 Model Equation")
print(f"Marks = {b0:.4f} + {b1:.4f}*x + {b2:.4f}*x^2")

# Step 8: Prediction
y_pred2 = model2.predict(X_poly2)

# Step 9: Calculate MSE and R2
mse2 = mean_squared_error(y, y_pred2)
r2_2 = r2_score(y, y_pred2)

print("Training MSE (Degree 2):", round(mse2, 4))
print("R2 Score (Degree 2):", round(r2_2, 4))

# Step 10: Plot graph
idx = np.argsort(X.values[:, 0])

plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X.values[idx, 0], y_pred2[idx],
         color='red', label='Degree 2')

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Polynomial Regression - Degree 2")
plt.legend()
plt.show()
```

```python

# Step 1: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Step 2: Read CSV file
df = pd.read_csv('regression.csv')

# Step 3: Select X and Y
X = df[['Hours_Studied']]
y = df['Marks']

# Step 4: Create degree 3 polynomial features
poly3 = PolynomialFeatures(degree=3, include_bias=False)
X_poly3 = poly3.fit_transform(X)

# Step 5: Train model
model3 = LinearRegression()
model3.fit(X_poly3, y)

# Step 6: Print model equation
b0 = model3.intercept_
b1, b2, b3 = model3.coef_

print("Degree 3 Model Equation")
print(f"Marks = {b0:.4f} + {b1:.4f}*x + {b2:.4f}*x^2 + {b3:.4f}*x^3")

# Step 7: Prediction
y_pred3 = model3.predict(X_poly3)

# Step 8: Calculate MSE and R2
mse3 = mean_squared_error(y, y_pred3)
r2_3 = r2_score(y, y_pred3)

print("Training MSE (Degree 3):", round(mse3, 4))
print("R2 Score (Degree 3):", round(r2_3, 4))

# Step 9: Plot graph
idx = np.argsort(X.values[:, 0])

plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X.values[idx, 0], y_pred3[idx],
         color='green', label='Degree 3')

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Polynomial Regression - Degree 3")
plt.legend()
plt.show()
```
