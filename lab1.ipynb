```python

import math

print(math.sqrt(25))
print(math.factorial(5))
print(math.pi)
print(math.e)
print(math.sin(math.radians(30)))


import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)

matrix = np.array([[1, 2], [3, 4]])
print(matrix)

print(np.zeros((2, 3)))
print(np.ones((3, 3)))
print(np.arange(0, 10, 2))

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)

print(np.mean(arr))
print(np.std(arr))


import pandas as pd

# Create Series
data = [10, 20, 30, 40]
series = pd.Series(data)
print(series)

# Series with custom index
series = pd.Series(
    [10, 20, 30, 40],
    index=['a', 'b', 'c', 'd']
)
print(series)

# Create DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Sal': [50000, 60000, 70000],
    'country': ['india', 'usa', 'uk']
}

df = pd.DataFrame(data)
print(df)


import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 35, 45, 50]

plt.plot(
    x, y,
    marker='o',
    linestyle='--',
    color='red',
    label='Line Graph'
)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Basic Line Graph')
plt.legend()
plt.grid(True)
plt.show()


import matplotlib.pyplot as plt

categories = ['A', 'B', 'C', 'D']
values = [10, 20, 30, 40]

plt.bar(categories, values, color='yellow')

plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Graph')

plt.show()


import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(1000)

plt.hist(
    data,
    bins=20,
    color='purple',
    edgecolor='black'
)

plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram Example')

plt.show()


import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(
    x, y,
    color='green',
    marker='*'
)

plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Scatter Plot')

plt.show()


import matplotlib.pyplot as plt

labels = ['Apple', 'Banana', 'Cherry', 'Date']
sizes = [20, 30, 35, 15]

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',
    colors=['blue', 'green', 'purple', 'yellow']
)

plt.title('Fruit Distribution')
plt.show()

```
